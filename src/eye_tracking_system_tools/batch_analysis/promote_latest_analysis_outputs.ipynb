# Promote Latest Analysis Outputs

Promote the newest heavy-analysis files from analysis subfolders (and the analysis root) to the analysis root so `BlockSync` picks them up by default.

**Promoted files:** brightness dict, jitter report dict, `final_sync_df.csv`.
**Behavior:** Top-level analysis files are included when choosing "most recent" by mtime; if the current top-level file is already newest, it is left in place.

Edit the parameters in the next cell, then run both cells.

## Parameters (edit and run)

```python
from pathlib import Path

# Experiment root (contains animal folders)
experiment_path = Path(r"D:\path\to\experiment")

# Animal name(s) to process
animals = ["PV_126", "PV_106"]

# Only consider analysis subfolders whose names start with this prefix.
# Use "" to include all subfolders.
source_prefix = "batch_analysis_output_"

# If True, only print what would be done; do not copy files
dry_run = True

# If True, do not backup existing root files before replacing
no_backup = False

# Optional: path to write a CSV report (None to skip)
output_csv = None  # e.g. Path("promote_report.csv")
```

## Run promotion

```python
from eye_tracking_system_tools.batch_analysis.promote_latest_analysis_outputs import run_promote

exit_code = run_promote(
    experiment_path,
    animals,
    source_prefix=source_prefix,
    no_backup=no_backup,
    dry_run=dry_run,
    output_csv=output_csv,
)
print(f"Exit code: {exit_code}")
```
