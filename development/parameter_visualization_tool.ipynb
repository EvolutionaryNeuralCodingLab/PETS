# Velocity / threshold tool (desktop GUI)

Minimal native window: **both-eye speed** traces + a **draggable horizontal threshold**.
Toggle **Degrees** / **Pixels**. No full saccade detector on drag — just the line and shading.

- Drag the black dashed line, or edit the Threshold spinbox
- **Auto-zoom** / **Full span** if the view looks empty
- **Save thr → YAML** writes `speed_threshold_deg_per_frame` (degrees mode only)
